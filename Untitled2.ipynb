```python

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Step 2: Read the text file
print("Step 2: Reading paragraph.txt...")

with open("paragraph.txt", "r", encoding="utf-8") as file:
    text = file.read().strip()

print("Text file loaded successfully!")
print("Total characters:", len(text))
print("\nSample text:")
print(text[:300])

# Step 3: Create character mappings
print("\nStep 3: Creating character mappings...")

chars = sorted(list(set(text)))

char_to_index = {char: i for i, char in enumerate(chars)}
index_to_char = {i: char for i, char in enumerate(chars)}

print("Unique characters:", len(chars))

# Step 4: Create input sequences and labels
print("\nStep 4: Creating input sequences...")

seq_length = 3
sequences = []
labels = []

for i in range(len(text) - seq_length):
    seq = text[i:i + seq_length]
    label = text[i + seq_length]

    sequences.append([char_to_index[char] for char in seq])
    labels.append(char_to_index[label])

X = np.array(sequences)
y = np.array(labels)

print("Total sequences:", len(sequences))
print("Input shape:", X.shape)
print("Labels shape:", y.shape)

# Step 5: Apply one-hot encoding
print("\nStep 5: Applying one-hot encoding...")

X_one_hot = tf.one_hot(X, len(chars))
y_one_hot = tf.one_hot(y, len(chars))

print("One-hot encoding completed!")

# Step 6: Build the LSTM model
print("\nStep 6: Building the LSTM model...")

model = Sequential()

model.add(LSTM(
    50,
    input_shape=(seq_length, len(chars))
))

model.add(Dense(len(chars), activation="softmax"))

model.summary()

# Step 7: Compile the model
print("\nStep 7: Compiling the model...")

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")

# Step 8: Train the model
print("\nStep 8: Training the model...")

model.fit(
    X_one_hot,
    y_one_hot,
    epochs=100,
    verbose=1
)

print("Model training completed!")

# Step 9: Generate new text
print("\nStep 9: Generating new text...")

start_seq = text[:seq_length]
generated_text = start_seq

for i in range(100):
    seq = generated_text[-seq_length:]

    x = np.array([
        [char_to_index[char] for char in seq]
    ])

    x_one_hot = tf.one_hot(x, len(chars))

    prediction = model.predict(x_one_hot, verbose=0)

    next_index = np.argmax(prediction[0])
    next_char = index_to_char[next_index]

    generated_text += next_char

# Step 10: Display output
print("\nStep 10: Final Output")

print("\nOriginal Text:")
print(text)

print("\nGenerated Text:")
print(generated_text)
```
